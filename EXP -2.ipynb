```python
import csv

# Read dataset
data = []
'''
Sky,AirTemp,Humidity,Wind,Water,Forecast,EnjoySport
Sunny,Warm,Normal,Strong,Warm,Same,Yes
Sunny,Warm,High,Strong,Warm,Same,Yes
Rainy,Cold,High,Strong,Warm,Change,No
Sunny,Warm,High,Strong,Cool,Change,Yes
'''
with open("candidate.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)

    for row in reader:
        data.append(row)

attributes = len(data[0]) - 1

# Most specific hypothesis
S = ['0'] * attributes

# Most general hypothesis
G = [['?'] * attributes]


def covers(h, x):
    for i in range(attributes):
        if h[i] != '?' and h[i] != x[i]:
            return False
    return True


for row in data:

    x = row[:-1]
    target = row[-1]

    if target == "Yes":

        # Remove general hypotheses that do not cover positive example
        G = [g for g in G if covers(g, x)]

        # Generalize S
        for i in range(attributes):

            if S[i] == '0':
                S[i] = x[i]

            elif S[i] != x[i]:
                S[i] = '?'

    else:

        # Remove general hypotheses that cover negative example
        new_G = []

        for g in G:

            if covers(g, x):

                for i in range(attributes):

                    if g[i] == '?':
                        h = g.copy()
                        h[i] = S[i]

                        if h[i] != '?' and h[i] != '0':
                            new_G.append(h)

            else:
                new_G.append(g)

        G = new_G


print("Specific Boundary S:")
print(S)

print("\nGeneral Boundary G:")
for g in G:
    print(g)
```
