```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error

# Dataset
X = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)

y = np.array([
    2, 5, 10, 17, 26, 37, 50, 65
])

# -------------------------
# Linear Regression
# -------------------------

linear_model = LinearRegression()

linear_model.fit(X, y)

linear_prediction = linear_model.predict(X)

linear_error = mean_squared_error(y, linear_prediction)


# -------------------------
# Polynomial Regression
# -------------------------

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

poly_model = LinearRegression()

poly_model.fit(X_poly, y)

poly_prediction = poly_model.predict(X_poly)

poly_error = mean_squared_error(y, poly_prediction)


# Results
print("Linear Regression")
print("MSE:", linear_error)

print("\nPolynomial Regression")
print("MSE:", poly_error)
```
